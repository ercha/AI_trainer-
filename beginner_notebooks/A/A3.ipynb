## 一、先认识术语

- **尺寸**：图像的宽度和高度；OpenCV resize 参数写 (宽,高)
- **颜色空间**：表示颜色的方法，常用 BGR、RGB、HSV、灰度
- **边界填充**：在图像四周补像素，避免卷积时图像边缘信息不足

resize 改变宽高；copyMakeBorder 扩展边缘；cvtColor 负责灰度、HSV、RGB 转换。与数组 shape 不同，resize 的尺寸参数是 (宽,高)。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import cv2, numpy as np
img = np.zeros((80, 120, 3), dtype=np.uint8)
small = cv2.resize(img, (60, 40))
gray = cv2.cvtColor(small, cv2.COLOR_BGR2GRAY)
frame = cv2.copyMakeBorder(gray, 5, 5, 5, 5, cv2.BORDER_REFLECT)
print(img.shape, small.shape, gray.shape, frame.shape)
````

## 三、逐句解释

1. img 是 80 行×120 列×3 通道。

2. resize(img,(60,40)) 产生 40 行×60 列×3 通道。

3. BGR2GRAY 后去掉颜色通道维；再分别补 5 像素得到 50 行×70 列。

### 额外提醒

- 图像数组 shape 常见为 (高,宽,通道)
- resize 的 dsize 则为 (宽,高)
- BGR2GRAY 后彩色通道维消失

## 四、重点函数与参数

### cv2.resize

- **写法**：`cv2.resize(src, dsize, interpolation=cv2.INTER_LINEAR)`
- **参数**：dsize：(宽,高)；interpolation：插值算法
- **返回**：返回缩放后的图像
- **特别注意**：shape 却按 (高,宽,通道) 输出

### cv2.cvtColor

- **写法**：`cv2.cvtColor(src, code)`
- **参数**：code：cv2.COLOR_BGR2GRAY / BGR2RGB / BGR2HSV
- **返回**：返回新颜色空间图像
- **特别注意**：使用 plt.imshow 显示 OpenCV 彩图前常需 BGR→RGB
