## 一、先认识术语

- **路径**：文件在电脑上的位置，可以是相对路径或绝对路径
- **解码**：将 PNG/JPEG 的压缩字节还原为可操作的像素数组
- **BGR**：OpenCV 彩色图默认的三个通道排列：蓝、绿、红

OpenCV 读入彩色图默认为 BGR。cv2.imread 失败会返回 None；中文路径可使用 np.fromfile + cv2.imdecode，保存使用 cv2.imencode(...).tofile。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from pathlib import Path
import cv2, numpy as np
src = Path("练习图片.png")
if src.exists():
    img = cv2.imdecode(np.fromfile(str(src), dtype=np.uint8), cv2.IMREAD_COLOR)
    if img is None:
        raise ValueError("无法解码图片")
    ok, buf = cv2.imencode(".png", img)
    if ok:
        buf.tofile("保存结果.png")
else:
    print("先在当前目录放入练习图片.png")
````

## 三、逐句解释

1. Path('练习图片.png') 指向当前工作目录的文件，而不是脚本本身所在目录。

2. np.fromfile 以字节形式读取，cv2.imdecode 按图片格式解码，中文路径也适用。

3. 保存时 cv2.imencode 先生成 PNG 字节，再调用 .tofile 写入中文路径。

### 额外提醒

- exists 先验证文件存在
- imdecode 负责解码图像字节
- imencode 得到可写入磁盘的图像字节

## 四、重点函数与参数

### cv2.imdecode

- **写法**：`cv2.imdecode(buf, flags)`
- **参数**：buf：uint8 字节数组；flags：cv2.IMREAD_COLOR / GRAYSCALE
- **返回**：成功返回图像数组；失败可能返回 None
- **特别注意**：在读取后检查 img is None

### cv2.imencode

- **写法**：`cv2.imencode(ext, img)`
- **参数**：ext：文件格式扩展名如 '.png'；img：数组
- **返回**：返回 (成功标志, 字节缓冲区)
- **特别注意**：一定先验证 ok 再执行 buf.tofile
