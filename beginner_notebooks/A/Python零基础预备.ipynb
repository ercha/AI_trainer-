## 3. 条件判断与循环

```python
scores=[59,75,88]
for score in scores:
    if score>=60:
        print(score,'及格')
    else:
        print(score,'不及格')

# TODO 3：写 for 循环，将及格人数赋给 passed_count
```

## 5. 文件路径、中文与异常

Path 用于构建路径；读写中文文件时指定 UTF-8；try/except 接住异常。


```python
from pathlib import Path
folder=Path('primer_output')
folder.mkdir(exist_ok=True)
p=folder/'成绩.txt'
p.write_text('通过\n',encoding='utf-8')
try:
    print(p.read_text(encoding='utf-8'))
except OSError as exc:
    print('读取失败:',exc)

# TODO 5：创建 primer_output/说明.txt 并写入一行中文
```

```python
import numpy as np
matrix=np.arange(12).reshape(3,4)
print(matrix)
print('形状:',matrix.shape,'第2行第3列:',matrix[1,2])
assert matrix[1,2]==6

# TODO 6：创建 2×3 零数组，将第 1 行第 2 列改为 255，用 assert 验证
```
