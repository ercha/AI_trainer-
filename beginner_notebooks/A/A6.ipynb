## 一、先认识术语

- **轮廓**：像素区域外边缘的点序列
- **外接矩形**：包含目标的最小轴对齐矩形，使用 x、y、w、h
- **交付物**：任务书要求最终提交的文件，如 CSV、PNG、模型

从掩膜提取轮廓后，可计算面积与外接框，并用 Matplotlib 比较处理前后。竞赛要核对实际文件是否保存成功。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import cv2, numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
img = np.zeros((80, 100), dtype=np.uint8)
cv2.rectangle(img, (20, 10), (70, 60), 255, -1)
contours, _ = cv2.findContours(img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
print("轮廓数量:", len(contours))
print("外接框:", cv2.boundingRect(contours[0]))
Path("output").mkdir(exist_ok=True)
plt.imshow(img, cmap="gray"); plt.title("mask"); plt.axis("off")
plt.savefig("output/preview.png", dpi=150); plt.close()
````

## 三、逐句解释

1. cv2.findContours 返回轮廓列表和层级信息。

2. cv2.boundingRect 的返回值代表左上角坐标和尺寸，不是右下角坐标。

3. plt.savefig 把图写入 output/，并用 Path.exists 检查。

### 额外提醒

- findContours 需要单通道二值图
- boundingRect 结果为 (x,y,w,h)
- savefig 后检查文件路径

## 四、重点函数与参数

### cv2.findContours

- **写法**：`cv2.findContours(image, mode, method)`
- **参数**：image：二值图；mode：RETR_EXTERNAL；method：CHAIN_APPROX_SIMPLE
- **返回**：返回 (contours, hierarchy)
- **特别注意**：OpenCV 4 的常见解包形式是 contours,_

### plt.savefig

- **写法**：`plt.savefig(fname, dpi=150, bbox_inches='tight')`
- **参数**：fname：目标路径；dpi：输出分辨率
- **返回**：写入 PNG 等文件
- **特别注意**：仅调用 plt.show 不代表已保存
