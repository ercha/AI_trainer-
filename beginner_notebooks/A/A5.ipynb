## 一、先认识术语

- **二值掩膜**：只有两种值的图，通常用 0 和 255
- **边缘**：像素亮度变化明显的轮廓线
- **开运算/闭运算**：先腐蚀后膨胀一般去小白点；先膨胀后腐蚀一般填小黑洞

阈值分割产生黑白掩膜；Canny 提取边缘；开运算去小噪点，闭运算填小孔洞。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import cv2, numpy as np
img = np.zeros((100, 100), dtype=np.uint8)
cv2.rectangle(img, (20,20), (80,80), 230, -1)
_, mask = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
edges = cv2.Canny(img, 80, 160)
kernel = np.ones((3,3), np.uint8)
clean = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
print(np.unique(clean), np.count_nonzero(edges))
````

## 三、逐句解释

1. 先构造灰度背景与亮色矩形。

2. threshold 函数返回两个值，真正的掩膜是第二个。

3. Canny 返回边缘图；morphologyEx 清理掩膜。

### 额外提醒

- threshold 返回 (阈值,结果图)
- Canny 的两级阈值控制边缘筛选
- morphologyEx 使用结构元素 kernel

## 四、重点函数与参数

### cv2.threshold

- **写法**：`cv2.threshold(src, thresh, maxval, type)`
- **参数**：thresh：分界阈值；maxval：常用 255；type：THRESH_BINARY
- **返回**：返回 (实际使用阈值, 图像)
- **特别注意**：_,mask=... 不能只写 mask=cv2.threshold(...)

### cv2.Canny

- **写法**：`cv2.Canny(image, threshold1, threshold2)`
- **参数**：threshold1/2：低/高梯度阈值
- **返回**：返回单通道边缘图
- **特别注意**：阈值不是图像亮度阈值；和 threshold 功能不同

## 五、运行结果核验
