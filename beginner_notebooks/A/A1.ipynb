## 一、先认识术语

- **像素**：一幅图最小的颜色采样单位；灰度图每个像素是一个 0～255 的数
- **数组**：按行列组织的数据容器；灰度图通常有两个维度
- **uint8**：无符号 8 位整数，范围 0～255；常用于存储普通图像

变量保存数据；print 输出；函数是可重复使用的操作。图像本质上是由像素组成的数组。先用最简单的 NumPy 数组理解灰度像素。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import numpy as np
pixels = np.array([[0, 128], [200, 255]], dtype=np.uint8)
print(pixels)
print("形状:", pixels.shape)
print("数据类型:", pixels.dtype)
````

## 三、逐句解释

1. 第 1 行导入 numpy 并起别名 np，之后调用 np.array 等函数。

2. np.array 把二维列表转换为形状 (2,2) 的数组；每个嵌套列表代表一行。

3. pixels.shape 返回高度与宽度；dtype 表示元素的存储类型。

### 额外提醒

- np.array 把列表变成数组
- shape 的两个数是行数、列数
- uint8 表示像素通常位于 0~255

## 四、重点函数与参数

### np.array

- **写法**：`np.array(object, dtype=None)`
- **参数**：object：输入列表；dtype：元素数据类型
- **返回**：返回 ndarray
- **特别注意**：uint8 不适合直接保存超过 255 的像素数值

### np.zeros

- **写法**：`np.zeros(shape, dtype=float)`
- **参数**：shape：各维长度；dtype：默认浮点类型
- **返回**：给定形状的全零数组
- **特别注意**：图像练习常明确写 dtype=np.uint8

## 五、运行结果核验

将显示 2×2 的整数数组、shape=(2,2)、dtype=uint8；它并没有图像文件。

> 结果可能因随机种子、包版本或操作系统略有差异。重点核对形状、字段类型、输出文件和运行是否报错。
