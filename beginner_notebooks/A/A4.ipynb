## 一、先认识术语

- **卷积核**：在局部像素范围做运算的小窗口，如 3×3、5×5
- **椒盐噪声**：随机出现的黑白孤立噪点
- **高斯权重**：中心像素权重较高、远处较低的平滑方式

滤波用于减少噪声。均值滤波取邻域均值；高斯滤波按距离加权；中值滤波抗椒盐噪声。常用核尺寸为奇数。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import numpy as np, cv2
img = np.full((60, 60), 120, dtype=np.uint8)
img[30, 30] = 255
mean = cv2.blur(img, (5, 5))
gauss = cv2.GaussianBlur(img, (5, 5), 0)
median = cv2.medianBlur(img, 5)
print(mean[30,30], gauss[30,30], median[30,30])
````

## 三、逐句解释

1. img = np.full 创建均匀灰色背景。

2. 设置中心像素为 255，模拟一个亮噪点。

3. 比较三种滤波下中心像素值，观察中值滤波消除孤立噪点的效果。

### 额外提醒

- blur 的核大小用元组
- GaussianBlur 第三个参数 0 表示自动计算 sigmaX
- medianBlur 第二个参数是奇数整数

## 四、重点函数与参数

### cv2.GaussianBlur

- **写法**：`cv2.GaussianBlur(src, ksize, sigmaX)`
- **参数**：ksize：奇数尺寸元组；sigmaX：X方向标准差，0 表示自动推导
- **返回**：返回平滑图
- **特别注意**：(5,5),0 是常见写法，不是『滤波关闭』

### cv2.medianBlur

- **写法**：`cv2.medianBlur(src, ksize)`
- **参数**：ksize：奇数整数，不是 (5,5) 元组
- **返回**：返回中值滤波图
- **特别注意**：若写成 (5,5) 会发生参数类型错误

## 五、运行结果核验

由于孤立亮噪点会被周围正常灰度覆盖，中值滤波中心结果常回到 120。
