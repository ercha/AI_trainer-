## 一、先认识术语

- **JSON**：用对象、数组存储字段和数据的文本格式
- **类型转换**：把价格字符串变成数字后才能用于乘法
- **GMV**：成交总额，这里用价格×数量求和作为示例

JSON 是一种常见数据交换格式，Pandas 可将对象列表转成表。先确定字段类型，再计算业务统计。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import pandas as pd
rows=[{"书名":"A","价格":"20","销量":2},{"书名":"B","价格":"30","销量":3}]
df=pd.DataFrame(rows)
df["价格"]=pd.to_numeric(df["价格"],errors="coerce")
df["销售额"]=df["价格"]*df["销量"]
print(df)
print("总销售额:",df["销售额"].sum())
````

## 三、逐句解释

1. records 是商品字典列表，每个字典作为 DataFrame 一行。

2. pd.to_numeric 将价格字符串转换为可计算数值。

3. df['销售额'] 创建新列，再调用 .sum() 计算总销售额。

### 额外提醒

- to_numeric(errors='coerce') 将非法数据转成 NaN
- 新建字段使用 df['列名']=...
- 指标前先核验原始字段含义

## 四、重点函数与参数

### pd.to_numeric

- **写法**：`pd.to_numeric(arg, errors='coerce')`
- **参数**：arg：输入列；errors='coerce' 使无效值变 NaN
- **返回**：数值 Series
- **特别注意**：非法值转 NaN 后仍要决定怎么填补或删除

### to_csv

- **写法**：`df.to_csv('cleaned.csv', index=False, encoding='utf-8-sig')`
- **参数**：index=False：不多写行索引；encoding：中文 Excel 兼容
- **返回**：写 CSV 文件
- **特别注意**：数据清洗后要确认文件真实存在
