## 一、先认识术语

- **Figure**：整张图像画布
- **Axes**：图中的绘图区，可设置坐标轴标题
- **dpi**：保存图像的像素密度设置，常用 150 或 300

项目图表要标明标题、坐标轴并保存 PNG。使用 plt.tight_layout 避免标签被遮挡；非英文字体可能需要按本机安装情况配置。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import matplotlib.pyplot as plt
from pathlib import Path
names=["A","B","C"]; values=[10,18,13]
plt.bar(names,values)
plt.title("Count by group")
plt.xlabel("Group"); plt.ylabel("Count")
plt.tight_layout()
Path("output").mkdir(exist_ok=True)
plt.savefig("output/chart.png",dpi=150)
plt.close()
print(Path("output/chart.png").exists())
````

## 三、逐句解释

1. plt.bar 生成按类别分布的柱形图。

2. title、xlabel、ylabel 增加解释，让阅卷人能理解内容。

3. tight_layout 防止挤压；savefig 写入 PNG；close 关闭图像释放资源。

### 额外提醒

- bar 绘制柱形图
- title/xlabel/ylabel 是基本标注
- savefig 保存到磁盘，close 释放资源

## 四、重点函数与参数

### plt.savefig

- **写法**：`plt.savefig(fname, dpi=150, bbox_inches='tight')`
- **参数**：fname：输出文件；dpi：图像清晰度；bbox_inches：边距调整
- **返回**：写入图像文件
- **特别注意**：保存前先 mkdir 创建目录

### plt.subplots

- **写法**：`fig,ax=plt.subplots(1,2,figsize=(10,4))`
- **参数**：行数、列数、画布尺寸
- **返回**：返回 Figure 与 Axes
- **特别注意**：多图建议使用 ax.plot、ax.bar、ax.set_title
