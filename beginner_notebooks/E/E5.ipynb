检索增强 RAG 的本质是『先查资料，再组织回答』。入门只需要 TF-IDF + 余弦相似度就能做完整检索，不强迫使用 GPU。

## 一、先认识术语

- **检索 Top-K**：按相关性选出最相近的 K 段资料
- **余弦相似度**：衡量两个向量方向接近程度的指标
- **延迟/吞吐量**：一次请求耗时，以及单位时间内处理多少请求

RAG 先从资料检索，再让模型回答。入门可先实现 TF-IDF 检索，不必接大模型。性能测试要清楚计算总耗时与平均延迟。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import time
docs=["图书馆周一开放","实验室需预约","食堂供应午餐"]
query="图书馆几点开放"
t0=time.perf_counter()
vec=TfidfVectorizer(analyzer="char",ngram_range=(2,3))
X=vec.fit_transform(docs)
q=vec.transform([query])
idx=cosine_similarity(q,X).argmax()
print("最相关:",docs[idx])
print("耗时(秒):",time.perf_counter()-t0)
````

## 三、逐句解释

1. TfidfVectorizer(analyzer='char') 将中文句子转成字符 n-gram 向量。

2. fit_transform 建立文档向量库；transform 将用户问题投影到相同词表。

3. cosine_similarity 计算相关性，argmax 选得分最高的文档，perf_counter 计时。

### 额外提醒

- 中文可使用字符 n-gram
- 相似度最高的文本并不保证答案一定正确
- 总耗时除以调用次数是平均延迟

## 四、重点函数与参数

### cosine_similarity

- **写法**：`cosine_similarity(query_vector, document_matrix)`
- **参数**：输入形状 (1,V) 与 (N,V)
- **返回**：返回形状 (1,N) 的相似度矩阵
- **特别注意**：得分高代表相关，不等于已验证事实

### time.perf_counter

- **写法**：`time.perf_counter()`
- **参数**：无参数，读取高精度计时器
- **返回**：返回秒数（相对时间）
- **特别注意**：延迟=t1-t0，吞吐量=总请求数/总耗时
