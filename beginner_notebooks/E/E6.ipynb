## 一、先认识术语

- **单元测试**：验证一个功能在给定输入下是否返回预期结果
- **回归测试**：修改代码后重复确认旧功能不受影响
- **README**：说明安装、运行、输入输出、测试结果的交付文档

一个可复现的项目应包含环境说明、启动命令、输入输出样例、测试结果和失败兜底。实践中先跑通最小闭环，再优化效果。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from pathlib import Path
required=["README.md","output/chart.png"]
for filename in required:
    path=Path(filename)
    print(("通过" if path.exists() else "缺失"),filename)
# 请按你自己的练习目录修改 required 列表
````

## 三、逐句解释

1. required 指定需要提交的文档和图像。

## 四、重点函数与参数

### assert

- **写法**：`assert condition, 'message'`
- **参数**：condition：必须满足的条件；message：失败提示
- **返回**：条件 False 时抛出 AssertionError
- **特别注意**：适用于测试断言，不能代替生产环境完整错误处理

### path.stat

- **写法**：`Path('output/chart.png').stat().st_size`
- **参数**：st_size：文件字节数
- **返回**：返回文件大小
- **特别注意**：存在但大小为零的文件不能算通过交付

## 五、运行结果核验

如果 README.md 存在但 output/chart.png 不存在，会显示『缺失 output/chart.png』。

> 结果可能因随机种子、包版本或操作系统略有差异。重点核对形状、字段类型、输出文件和运行是否报错。
