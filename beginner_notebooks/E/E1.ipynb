## 四、重点函数与参数

### Path.exists

- **写法**：`Path(path).exists()`
- **参数**：path：文件或目录的位置
- **返回**：True/False
- **特别注意**：存在不代表内容正确，还应检查大小/列名等

### Path.mkdir

- **写法**：`Path('output').mkdir(parents=True, exist_ok=True)`
- **参数**：parents：创建父目录；exist_ok：已存在不报错
- **返回**：创建目录
- **特别注意**：建议在写文件前调用
