## 一、先认识术语

- **API**：程序之间约定的调用接口
- **环境变量**：保存地址、模型名、密钥等配置的操作系统变量
- **超时与异常处理**：网络慢、服务失败时不能让批处理无限等待或直接崩溃

本地模型用 predict；远程大模型 API 用 HTTP 请求，需配置地址、模型名和密钥。不要把私钥写入项目代码；使用环境变量，并处理断网/响应异常。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import os
base=os.getenv("OPENAI_BASE_URL","")
model=os.getenv("OPENAI_MODEL","")
key=os.getenv("OPENAI_API_KEY","")
if not all([base,model,key]):
    print("未配置 API，先做离线数据分析")
else:
    print("已检测到配置，可进行联网调用（此示例不发送请求）")
````

## 三、逐句解释

1. os.getenv 从环境变量读取地址、模型名、密钥；缺失时返回空字符串。

2. all([...]) 检查三者是否齐全，缺失则走离线路径。

3. 实际调用 requests.post 前必须验证地址，并使用 timeout；返回 JSON 要检查 choices 等字段。

### 额外提醒

- API 密钥从环境变量获取
- 未配置时提供离线替代路径
- 联网调用要考虑超时和响应解析

## 四、重点函数与参数

### os.getenv

- **写法**：`os.getenv('OPENAI_BASE_URL', '')`
- **参数**：第一个参数：变量名；第二个参数：缺失时默认值
- **返回**：返回配置字符串
- **特别注意**：不要把 API Key 写入版本库

### requests.post

- **写法**：`requests.post(url, headers=..., json=..., timeout=60)`
- **参数**：json：请求体；headers：认证；timeout：最大等待时间
- **返回**：HTTP Response
- **特别注意**：先检查 status_code/raise_for_status 再解析 JSON
