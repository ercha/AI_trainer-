## 第一步：运行完整示例
先观察运行结果，再逐句阅读注释。

```python
from pathlib import Path
import numpy as np
import cv2
import matplotlib.pyplot as plt

out = Path("beginner_outputs")
out.mkdir(exist_ok=True)
img = np.zeros((100, 160, 3), dtype=np.uint8)
cv2.rectangle(img, (30, 20), (130, 80), (0, 255, 0), -1)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)
cv2.imencode(".png", blur)[1].tofile(str(out / "A_高斯滤波.png"))
plt.imshow(blur, cmap="gray")
plt.title("Gaussian blur")
plt.axis("off")
plt.show()
print("图像形状:", blur.shape)
```

## 第二步：自己动手
将高斯核改为 3×3，并额外保存一张 Canny 边缘图。

提示：可回到零基础网页课程查阅概念及参考写法。

```python
# TODO：生成 Canny 边缘，并把它保存到 beginner_outputs/A_edges.png
edges = cv2.Canny(gray, 80, 160)
cv2.imwrite(str(out / "A_edges.png"), edges)
print("边缘像素数:", np.count_nonzero(edges))
```
