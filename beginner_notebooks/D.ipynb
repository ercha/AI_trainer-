## 第一步：运行完整示例
先观察运行结果，再逐句阅读注释。

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

texts = ["足球 比赛 赢了", "球队 进球 领先", "软件 电脑 更新", "算法 模型 训练"]
labels = [0, 0, 1, 1]  # 0 体育，1 科技
model = Pipeline([
    ("tfidf", TfidfVectorizer(analyzer="char", ngram_range=(2, 3))),
    ("clf", LogisticRegression())
])
model.fit(texts, labels)
print("体育/科技预测:", model.predict(["今天球队进球", "软件算法升级"]))
```

## 第二步：自己动手
新增两条训练文本，重新训练模型并预测。

提示：可回到零基础网页课程查阅概念及参考写法。

```python
texts += ["篮球 运动 比赛", "人工智能 算法 科技"]
labels += [0, 1]
model.fit(texts, labels)
print("新预测:", model.predict(["篮球比赛", "人工智能软件"]))
```
