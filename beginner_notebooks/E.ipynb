## 第一步：运行完整示例
先观察运行结果，再逐句阅读注释。

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

out = Path("beginner_outputs")
out.mkdir(exist_ok=True)
records = [
    {"商品":"A","价格":12,"数量":3},
    {"商品":"B","价格":20,"数量":2},
    {"商品":"C","价格":15,"数量":4}
]
(out/"E_input.json").write_text(json.dumps(records,ensure_ascii=False),encoding="utf-8")
df = pd.read_json(out/"E_input.json")
df["销售额"] = df["价格"] * df["数量"]
df.to_csv(out/"E_result.csv", index=False, encoding="utf-8-sig")
print("GMV:", df["销售额"].sum())
plt.bar(df["商品"], df["销售额"])
plt.title("Sales by item")
plt.xlabel("Item"); plt.ylabel("Sales")
plt.tight_layout()
plt.savefig(out/"E_chart.png",dpi=150)
plt.close()
```

## 第二步：自己动手
把总销售额写入 Markdown 报告，并检查所有输出文件。

提示：可回到零基础网页课程查阅概念及参考写法。

```python
total = df["销售额"].sum()
(out/"E_report.md").write_text(
    f"# 简易商品报告\n\n总销售额：{total:.2f}\n\n已生成 CSV 与 PNG 图表。\n",
    encoding="utf-8"
)
required = ["E_input.json","E_result.csv","E_chart.png","E_report.md"]
for filename in required:
    print(filename, (out/filename).exists())
```
