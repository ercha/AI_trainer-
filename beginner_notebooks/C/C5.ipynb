## 一、先认识术语

- **state_dict**：模型可学习参数等状态的字典
- **eval 模式**：预测/验证模式，影响 Dropout 和 BatchNorm 等层
- **no_grad**：暂停构建梯度计算图，节省验证内存

训练使用 model.train；验证使用 model.eval 搭配 torch.no_grad。保存 state_dict 可复用模型权重，加载时结构必须匹配。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import torch
from torch import nn
net=nn.Linear(3,2)
net.eval()
with torch.no_grad():
    pred=net(torch.randn(4,3))
torch.save(net.state_dict(),"demo.pt")
loaded=nn.Linear(3,2)
loaded.load_state_dict(torch.load("demo.pt",map_location="cpu",weights_only=True))
print(pred.shape)
````

## 三、逐句解释

1. nn.Linear(3,2) 创建最小模型。

2. net.eval(); with torch.no_grad() 在无梯度记录下预测。

3. torch.save 保存参数字典，重新建同结构网络后 load_state_dict 恢复。

### 额外提醒

- eval 会影响 Dropout/BatchNorm 等层
- no_grad 避免记录验证阶段计算图
- 重新加载时创建相同结构的模型

## 四、重点函数与参数

### torch.save

- **写法**：`torch.save(obj, f)`
- **参数**：obj：通常是 model.state_dict()；f：保存路径
- **返回**：写入权重文件
- **特别注意**：保存 state_dict 更方便部署及版本控制

### torch.load

- **写法**：`torch.load(file, map_location='cpu', weights_only=True)`
- **参数**：map_location：目标设备；weights_only：安全加载常规权重
- **返回**：返回状态字典
- **特别注意**：只加载来自可信来源的模型文件
