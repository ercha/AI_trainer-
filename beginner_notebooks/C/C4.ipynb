## 一、先认识术语

- **NCHW**：批量 N、通道 C、高 H、宽 W 的 Tensor 维序
- **卷积**：用卷积核在图像上提取局部特征
- **池化**：缩小图像空间尺度；最大池化保留局部最强响应

Conv2d 处理二维图像；常见输入维度是 NCHW（批量、通道、高、宽）。MaxPool2d 会缩小空间尺寸，分类头常把特征图展平。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import torch
from torch import nn
x=torch.randn(2,1,20,20)
net=nn.Sequential(nn.Conv2d(1,8,3),nn.ReLU(),nn.MaxPool2d(2),nn.Flatten(),nn.Linear(8*9*9,10))
y=net(x)
print("输出形状:",y.shape)
````

## 三、逐句解释

1. 输入形状 (2,1,20,20)。

2. Conv2d(1,8,3) 默认无 padding：20-3+1=18。

3. MaxPool2d(2) 使 18×18→9×9；Flatten 得 8×9×9=648，Linear 映射到 10 类。

### 额外提醒

- 无 padding 的 3×3 卷积让 20×20 变成 18×18
- 2×2 池化后变成 9×9
- Linear 的输入数为 8×9×9

## 四、重点函数与参数

### nn.Conv2d

- **写法**：`nn.Conv2d(in_channels, out_channels, kernel_size, stride=1, padding=0)`
- **参数**：通道数、核大小、步长、填充
- **返回**：卷积后特征图
- **特别注意**：输出尺寸 floor((H+2P-K)/S)+1（不考虑膨胀）

### nn.Flatten

- **写法**：`nn.Flatten(start_dim=1)`
- **参数**：start_dim=1：保留 batch 维
- **返回**：将 (N,C,H,W) 变 (N,C×H×W)
- **特别注意**：接 Linear 前须计算展平后的维度
