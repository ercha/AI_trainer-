## 一、先认识术语

- **epoch**：把整个训练数据遍历一遍
- **梯度**：损失对参数变化率的估计，决定更新方向
- **优化器**：根据梯度调整网络参数的算法，如 SGD/Adam

一次训练迭代包含：清空梯度 zero_grad，前向预测，计算损失，backward 反向传播，step 更新参数。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import torch
from torch import nn
x=torch.tensor([[0.],[1.],[2.],[3.]])
y=2*x+1
net=nn.Linear(1,1)
opt=torch.optim.SGD(net.parameters(),lr=0.05)
for epoch in range(100):
    opt.zero_grad()
    loss=nn.MSELoss()(net(x),y)
    loss.backward()
    opt.step()
print("最终损失:",loss.item())
````

## 三、逐句解释

1. x 与 y 构造线性映射 y=2x+1 的训练数据。

2. opt.zero_grad(); loss=...; loss.backward(); opt.step() 是固定循环。

3. 随着 epoch 增加，loss 一般逐渐降低。

### 额外提醒

- zero_grad 避免上一步梯度累积
- backward 计算各参数梯度
- step 实际更新模型参数

## 四、重点函数与参数

### optimizer.zero_grad

- **写法**：`optimizer.zero_grad()`
- **参数**：清零上一轮梯度
- **返回**：无返回值
- **特别注意**：梯度默认累积；训练前应清零

### loss.backward

- **写法**：`loss.backward()`
- **参数**：对当前标量损失反向传播
- **返回**：写入参数 .grad
- **特别注意**：这一步并不直接更新参数；随后还要 optimizer.step
