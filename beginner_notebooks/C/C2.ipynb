## 一、先认识术语

- **神经元/层**：nn.Linear 定义可学习的线性变换
- **logits**：分类模型最后一层未经 softmax 的原始得分
- **交叉熵**：分类训练中衡量预测分数与正确类别差异的损失

nn.Linear 表示线性层；前向传播给出预测值；loss 比较预测与真实目标。分类用 CrossEntropyLoss 时应输入 logits，不需要提前 softmax。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import torch
from torch import nn
x=torch.randn(4,3)
y=torch.tensor([0,1,0,1])
net=nn.Sequential(nn.Linear(3,8),nn.ReLU(),nn.Linear(8,2))
out=net(x)
loss=nn.CrossEntropyLoss()(out,y)
print(out.shape,loss.item())
````

## 三、逐句解释

1. x=torch.randn(4,3) 模拟 4 个 3 维样本。

2. nn.Sequential 将 Linear、ReLU 和 Linear 按顺序串接。

3. 输出 logits=(4,2)，CrossEntropyLoss 读取每行的正确类别索引。

### 额外提醒

- 4 表示一批 4 个样本
- 输出最后一维 2 是两类 logits
- CrossEntropyLoss 接受整数类别标签

## 四、重点函数与参数

### nn.Linear

- **写法**：`nn.Linear(in_features, out_features)`
- **参数**：in_features：输入特征数；out_features：输出特征数
- **返回**：带可学习参数的线性层
- **特别注意**：上一层的 out_features 必须等于下一层 in_features

### nn.CrossEntropyLoss

- **写法**：`nn.CrossEntropyLoss()(logits, class_index)`
- **参数**：logits：形状 (N,C)；class_index：形状 (N,)，整数类别
- **返回**：标量损失 Tensor
- **特别注意**：不要在 logits 前额外调用 softmax
