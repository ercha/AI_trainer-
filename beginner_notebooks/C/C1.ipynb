## 一、先认识术语

- **Tensor**：PyTorch 的多维数值数组，支持梯度计算
- **device**：张量存储及计算的位置，例如 cpu 或 cuda
- **batch**：一次交给模型处理的样本组

Tensor 类似 NumPy 数组，但支持自动求导与 GPU 运算。训练前用 torch.cuda.is_available 检查 NVIDIA CUDA；CPU 同样能完成基础练习。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import torch
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
x=torch.tensor([[1.,2.],[3.,4.]],device=device)
print("设备:",device)
print("形状:",x.shape)
print(x.mean())
````

## 三、逐句解释

1. 先检查 torch.cuda.is_available 是否能使用 NVIDIA CUDA。

2. torch.device 用于构建实际设备，torch.tensor(..., device=device) 在指定设备创建张量。

3. x.mean() 返回张量元素均值，print(x.shape) 检查矩阵行列。

### 额外提醒

- torch.tensor 创建张量
- device 决定张量在哪运行
- print(x.shape) 检查维度

## 四、重点函数与参数

### torch.cuda.is_available

- **写法**：`torch.cuda.is_available()`
- **参数**：无参数；检查 PyTorch CUDA 能否实际使用
- **返回**：布尔值 True/False
- **特别注意**：nvidia-smi 识别到显卡，不代表对应 PyTorch wheel 可用

### torch.zeros

- **写法**：`torch.zeros(shape, device=device, dtype=torch.float32)`
- **参数**：shape：维度；device：运算设备；dtype：元素类型
- **返回**：全零张量
- **特别注意**：需要和模型参数放同一个 device
