## 一、先认识术语

- **连续值**：例如价格、温度，通常有很多可能的数值
- **RMSE**：均方误差的平方根，单位与预测目标相同
- **R²**：回归评价指标，1 代表完美拟合，也可能为负

回归预测连续数值。常见模型有线性回归、决策树和随机森林。RMSE 越小通常表示预测误差越小。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
X,y=make_regression(n_samples=120,n_features=3,noise=8,random_state=42)
Xtr,Xte,ytr,yte=train_test_split(X,y,random_state=42)
model=LinearRegression().fit(Xtr,ytr)
pred=model.predict(Xte)
print("RMSE:",np.sqrt(mean_squared_error(yte,pred)))
print("R2:",r2_score(yte,pred))
````

## 三、逐句解释

1. make_regression 生成虚拟连续数值数据，无需外部文件。

2. LinearRegression 拟合训练集，预测测试集。

3. mean_squared_error 求平方误差平均值，np.sqrt 再开方得到 RMSE。

### 额外提醒

- 回归预测值不是类别编号
- RMSE 是均方误差的平方根
- R² 越高通常拟合越好

## 四、重点函数与参数

### mean_squared_error

- **写法**：`mean_squared_error(y_true, y_pred)`
- **参数**：y_true：真实数值；y_pred：预测数值
- **返回**：返回均方误差 MSE
- **特别注意**：RMSE=np.sqrt(MSE)

### r2_score

- **写法**：`r2_score(y_true,y_pred)`
- **参数**：按测试集真实值和预测值比较
- **返回**：返回决定系数 R²
- **特别注意**：R² 可能小于 0，不能机械限定在 0～1

## 五、运行结果核验
