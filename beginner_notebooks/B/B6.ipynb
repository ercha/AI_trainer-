## 一、先认识术语

- **无监督学习**：训练时没有标准 y 标签，算法只看 X 的分布
- **聚类中心**：KMeans 每个分组的代表性中心
- **轮廓系数**：描述样本与本簇及其他簇的相对距离，可用于比较 k

聚类是不依赖标准类别标签的分组。KMeans 需要设置簇数 k；轮廓系数衡量簇内紧密与簇间分离。joblib 可保存训练好的模型。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import joblib
X,_=make_blobs(n_samples=100,centers=3,random_state=42)
model=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=model.fit_predict(X)
print("轮廓系数:",silhouette_score(X,labels))
joblib.dump(model,"kmeans_model.pkl")
````

## 三、逐句解释

1. make_blobs 生成 100 个分为 3 簇的二维点。

2. fit_predict(X) 先训练 KMeans，再对每个样本给出聚类编号。

3. silhouette_score 计算结果；joblib.dump 保存模型到 pkl 文件。

### 额外提醒

- fit_predict 会训练并给出聚类编号
- 没有真实标签也能评价聚类结构
- 保存 .pkl 方便后续加载

## 四、重点函数与参数

### KMeans

- **写法**：`KMeans(n_clusters=3, random_state=42, n_init=10)`
- **参数**：n_clusters：簇数；n_init：初始化尝试次数
- **返回**：返回聚类器
- **特别注意**：聚类编号 0/1/2 不等同于真实业务标签

### silhouette_score

- **写法**：`silhouette_score(X, labels)`
- **参数**：X：原始特征；labels：各样本的簇编号
- **返回**：返回 -1～1 的系数（在适用条件下）
- **特别注意**：需至少两个有效簇，单簇会报错
