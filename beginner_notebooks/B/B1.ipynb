## 一、先认识术语

- **DataFrame**：带行和列标签的二维数据表
- **特征 X**：用于预测的输入信息，可以是一列或多列
- **目标 y**：已有的正确标签或连续数值，训练时用于比较

一行是一个样本，一列是一个字段。X 表示用于预测的特征，y 表示标准答案/目标。DataFrame 是 Pandas 的二维表格。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
import pandas as pd
df = pd.DataFrame({"年龄":[18,25,30],"学习小时":[2,5,8],"通过":[0,1,1]})
print(df.head())
X = df[["年龄","学习小时"]]
y = df["通过"]
print(X.shape, y.shape)
````

## 三、逐句解释

1. pd.DataFrame 用字典构造三行数据；字典键自动成为列名。

2. df[['年龄','学习小时']] 返回形状 (3,2) 的特征表。

3. df['通过'] 返回形状 (3,) 的单列目标 Series。

### 额外提醒

- 双中括号取得二维特征表
- 单中括号取得目标列
- 先检查 head、shape、dtypes

## 四、重点函数与参数

### pd.read_csv

- **写法**：`pd.read_csv(filepath_or_buffer, encoding='utf-8')`
- **参数**：filepath_or_buffer：CSV 文件路径；encoding：编码
- **返回**：返回 DataFrame
- **特别注意**：中文 Excel 导出 CSV 可能是 utf-8-sig，需要检查

### DataFrame.drop

- **写法**：`df.drop(columns=['target'])`
- **参数**：columns：要移除的列名列表
- **返回**：返回删列后的表（默认不修改原表）
- **特别注意**：X=df.drop(columns=['target']); y=df['target']

## 五、运行结果核验

X 的 shape 为 (3,2)，y 的 shape 为 (3,)。

> 结果可能因随机种子、包版本或操作系统略有差异。重点核对形状、字段类型、输出文件和运行是否报错。
