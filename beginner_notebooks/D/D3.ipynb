## 一、先认识术语

- **文本标签**：例如体育=0、科技=1
- **向量化器**：把新文本用与训练时一致的词表映射到数字
- **误分类**：模型预测标签与真实标签不一致的样本

文本分类流程：标签、训练集、TF-IDF、分类器、预测和指标。Pipeline 能防止训练/测试词表不一致。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
texts=["体育 足球 比赛","体育 球队 进球","科技 电脑 软件","科技 算法 模型"]
y=[0,0,1,1]
pipe=Pipeline([("tfidf",TfidfVectorizer()),("clf",LogisticRegression())])
pipe.fit(texts,y)
print(pipe.predict(["足球 进球","电脑 算法"]))
````

## 三、逐句解释

1. texts 和 y 定义四条简短训练语句及类别。

2. TfidfVectorizer 和 LogisticRegression 用 Pipeline 顺序连接。

3. pipe.fit 训练；predict 对『足球进球』『电脑算法』输出类别。

### 额外提醒

- 整个 Pipeline 只需一次 fit
- predict 自动使用同一个向量器
- 真实项目应划分测试集并计算 F1

## 四、重点函数与参数

### Pipeline

- **写法**：`Pipeline([('tfidf', TfidfVectorizer()),('clf', LogisticRegression())])`
- **参数**：一个向量化步骤 + 一个分类器步骤
- **返回**：返回统一预测器
- **特别注意**：训练/预测会自动复用相同词表

### classification_report

- **写法**：`classification_report(y_true,y_pred,zero_division=0)`
- **参数**：输入真实/预测标签；zero_division 控制分母为 0 的处理
- **返回**：返回文本形式指标报告
- **特别注意**：正式评价必须划分独立测试集
