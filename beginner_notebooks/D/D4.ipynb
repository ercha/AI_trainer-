## 一、先认识术语

- **主题模型**：从多篇文档挖掘潜在词语分布的无监督模型
- **LDA**：基于概率分布的主题模型，这里是 LatentDirichletAllocation
- **命名实体**：人、地、机构等专有名称，NER 任务要识别其类型

Word2Vec 使词具有数值向量，能够测量相似性。LDA 通过词分布发现文档主题。NER 则提取人名、地名、组织等实体。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
texts=["足球 运动 球队","球队 足球 比赛","电脑 软件 算法","科技 算法 软件"]
vec=CountVectorizer()
X=vec.fit_transform(texts)
lda=LatentDirichletAllocation(n_components=2,random_state=42)
print("主题概率 shape:",lda.fit_transform(X).shape)
````

## 三、逐句解释

1. CountVectorizer 把每条空格分词的句子转为词频矩阵。

2. LDA 设置 n_components=2，拟合这组小语料的主题分布。

3. fit_transform 的输出每行包含该文档在 2 个主题上的概率。

### 额外提醒

- Word2Vec 任务通常要先分词
- LDA 是无监督主题模型
- 实体抽取返回结构化名称或位置

## 四、重点函数与参数

### CountVectorizer

- **写法**：`CountVectorizer(max_features=1500)`
- **参数**：max_features：限制词表规模
- **返回**：返回文本词频矩阵
- **特别注意**：与 TF-IDF 不同，词频不会先做 TF-IDF 权重

### LatentDirichletAllocation

- **写法**：`LatentDirichletAllocation(n_components=2, random_state=42)`
- **参数**：n_components：主题数量
- **返回**：返回可拟合 LDA 对象
- **特别注意**：主题编号是无监督结果，不代表人工标注类别
