把词变成数字是传统 NLP 的第一步。TF-IDF 并不是深度学习，而是一种词语特征权重算法。

> 零基础学习顺序：先读术语和说明 → 运行示例 → 理解代码 → 完成练习 → 概念自测。按 Shift+Enter 运行单元格。

## 一、先认识术语

- **词袋**：统计词是否出现及出现多少次，不考虑顺序
- **TF-IDF**：同时考虑词在本文中的频率与跨文档稀有程度
- **稀疏矩阵**：大多数位置为 0 的矩阵，能节省存储空间

模型不能直接计算自然语言，需要转成数值向量。CountVectorizer 统计词频；TfidfVectorizer 提高具有区分度词语的权重。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
from sklearn.feature_extraction.text import TfidfVectorizer
texts=["体育 比赛 足球","科技 人工智能 计算机","体育 足球 球队"]
vec=TfidfVectorizer()
X=vec.fit_transform(texts)
print(vec.get_feature_names_out())
print(X.shape)
````

## 三、逐句解释

1. TfidfVectorizer 在训练文本中建立词表。

2. fit_transform 同时学词表并把文章转换为矩阵。

3. get_feature_names_out 显示词表；X.shape 是 (文章数,词数)。

### 额外提醒

- fit 学习词表
- transform 将文本转成向量
- shape=(文本数,词表长度)

## 四、重点函数与参数

### TfidfVectorizer

- **写法**：`TfidfVectorizer(max_features=None, ngram_range=(1,1))`
- **参数**：max_features：最大词数；ngram_range：考虑的 n-gram 范围
- **返回**：返回向量器；fit_transform 返回稀疏矩阵
- **特别注意**：中文已按空格分好词时可直接使用默认分词器

### fit_transform

- **写法**：`vectorizer.fit_transform(texts)`
- **参数**：texts：字符串列表
- **返回**：返回数值特征矩阵
- **特别注意**：新文本只需 transform，不能重新 fit

## 五、运行结果核验

词表会包含足球、体育、计算机等词；矩阵行数等于文本数 3。
