## 一、先认识术语

- **字符串 str**：Python 处理文本的类型，支持切片、strip、replace
- **分词**：把一句话拆成有意义的词列表
- **停用词**：如『的、了』等常在特定任务中不提供明显区分信息的词

文本首先是字符串。strip 去掉首尾空白；replace 替换；jieba 可把中文句子分为词。基础课程不必先懂 Transformer。

## 二、运行示例

逐格执行并查看输出。这里的代码使用本地/合成数据，遇到缺失输入会有提示。不要直接跳过基础代码。

````python
text="  人工智能正在改变学习。  "
clean=text.strip().replace("。","")
print(clean)
try:
    import jieba
    print(list(jieba.cut(clean)))
except ImportError:
    print("需安装 competition_advanced_requirements.txt")
````

## 三、逐句解释

1. strip() 删除原句前后空格，replace('。','') 移除句末符号。

2. jieba.cut 返回可迭代对象，因此用 list 包装查看分词结果。

3. 没有 jieba 时例子只提示安装额外依赖，不中断其它模块。

### 额外提醒

- strip() 处理空白
- replace() 替换指定文本
- 中文分词结果是词列表

## 四、重点函数与参数

### jieba.cut

- **写法**：`jieba.cut(sentence, cut_all=False)`
- **参数**：sentence：中文句子；cut_all=False：精确模式
- **返回**：返回可迭代分词结果
- **特别注意**：分词结果可能受词典影响，需实际检查

### str.strip

- **写法**：`text.strip()`
- **参数**：默认清除首尾空白字符
- **返回**：返回新字符串
- **特别注意**：不会改变原始字符串变量，需重新赋值

## 五、运行结果核验

会输出没有首尾空格和句号的文本，并列出若干中文词。

> 结果可能因随机种子、包版本或操作系统略有差异。重点核对形状、字段类型、输出文件和运行是否报错。
