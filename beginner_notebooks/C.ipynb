## 第一步：运行完整示例
先观察运行结果，再逐句阅读注释。

```python
import torch
from torch import nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
x = torch.arange(0, 10, dtype=torch.float32).reshape(-1, 1).to(device)
y = 2*x + 1
model = nn.Linear(1, 1).to(device)
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
for epoch in range(300):
    optimizer.zero_grad()
    pred = model(x)
    loss = nn.MSELoss()(pred, y)
    loss.backward()
    optimizer.step()
print("运行设备:", device)
print("最终损失:", loss.item())
print("预测 x=5:", model(torch.tensor([[5.]], device=device)).item())
```

## 第二步：自己动手
保存和重新加载模型权重，验证预测是否一致。

提示：可回到零基础网页课程查阅概念及参考写法。

```python
from pathlib import Path
Path("beginner_outputs").mkdir(exist_ok=True)
torch.save(model.cpu().state_dict(), "beginner_outputs/C_linear.pt")
restored = nn.Linear(1, 1)
restored.load_state_dict(torch.load(
    "beginner_outputs/C_linear.pt", map_location="cpu", weights_only=True
))
restored.eval()
with torch.no_grad():
    print("加载后预测:", restored(torch.tensor([[5.]])).item())
```
